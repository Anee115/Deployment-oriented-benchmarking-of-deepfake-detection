```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ─────────────────────────────────────────
# DATA
# ─────────────────────────────────────────

datasets = [
    "DFC",
    "Celeb-DF (V2)",
    "FF++",
    "WildDeepfake"
]

models = [
    'XceptionNet',
    'ResNet-50',
    'MesoNet',
    'Inception-v3',
    'EfficientNetV2-B0',
    'ViT-16',
    'ViT-32',
    'Swin V2',
    'CrossViT',
    'MaxViT',
    'DINOv2'
]

# ─────────────────────────────────────────
# PEAK GPU MEMORY (MB)
# Mean values over five runs
# Rows = datasets
# Columns = models
# ─────────────────────────────────────────

gpu_usage = np.array([

    # DFC
    [
        7119.4, 7005.4, 7183.2, 7000.6, 6983.0,
        3432.9, 3412.8, 3258.8, 2627.3, 7342.6, 3668.4
    ],

    # Celeb-DF (V2)
    [
        7119.4, 7003.7, 7183.2, 7000.6, 6982.8,
        2557.1, 3412.2, 3249.1, 2615.3, 7395.1, 3947.3
    ],

    # FF++
    [
        7119.4, 7005.4, 7183.2, 7000.6, 6983.0,
        3436.2, 3570.2, 3249.2, 2615.3, 7342.3, 3681.8
    ],

    # WildDeepfake
    [
        7119.4, 7000.5, 7183.2, 7000.6, 6958.7,
        2636.7, 6455.0, 2992.6, 2723.4, 7238.6, 3666.4
    ]
])


# ─────────────────────────────────────────
# PEAK RAM MEMORY (MB)
# Mean values over five runs
# ─────────────────────────────────────────

ram_usage = np.array([

    # DFC
    [
        6409.1, 6442.7, 6302.2, 6507.5, 6531.9,
        4613.7, 6798.4, 6834.4, 6884.1, 6997.5, 6822.6
    ],

    # Celeb-DF (V2)
    [
        7353.1, 7392.3, 7270.6, 7446.1, 7478.3,
        904.5, 7500.1, 2378.6, 7528.8, 2383.3, 1980.6
    ],

    # FF++
    [
        5345.3, 5369.3, 5259.9, 5449.0, 5463.1,
        5762.0, 5783.0, 5799.1, 5807.4, 6207.3, 5787.3
    ],

    # WildDeepfake
    [
        9713.2, 4068.8, 9703.6, 8516.2, 9819.9,
        8801.3, 5150.6, 5318.0, 2291.1, 4810.0, 5096.6
    ]
])


# ─────────────────────────────────────────
# RESOURCE INDEX
#
# Min-max normalization is performed
# separately within each dataset.
#
# Resource Index =
# 0.5 × normalized GPU +
# 0.5 × normalized RAM
#
# LOWER = BETTER
# ─────────────────────────────────────────

def minmax_norm(arr):

    minimum = arr.min(
        axis=1,
        keepdims=True
    )

    maximum = arr.max(
        axis=1,
        keepdims=True
    )

    return (
        (arr - minimum) /
        (maximum - minimum + 1e-9)
    )


gpu_n = minmax_norm(
    gpu_usage
)

ram_n = minmax_norm(
    ram_usage
)

resource_index = (
    0.5 * gpu_n +
    0.5 * ram_n
)


# ─────────────────────────────────────────
# LONG DATAFRAME
# ─────────────────────────────────────────

rows = []

for di, d in enumerate(datasets):

    for mi, m in enumerate(models):

        rows.append({

            "Dataset":
            d,

            "Model":
            m,

            "Peak_GPU_MB":
            gpu_usage[di, mi],

            "Peak_RAM_MB":
            ram_usage[di, mi],

            "Normalized_GPU":
            gpu_n[di, mi],

            "Normalized_RAM":
            ram_n[di, mi],

            "ResourceIndex":
            resource_index[di, mi]

        })


df_idx = pd.DataFrame(
    rows
)


# ─────────────────────────────────────────
# COLOR PALETTE
# ─────────────────────────────────────────

colors = [
    '#C0392B',  # XceptionNet
    '#E67E22',  # ResNet-50
    '#8B6914',  # MesoNet
    '#27AE60',  # Inception-v3
    '#1E8449',  # EfficientNetB0
    '#1A5276',  # ViT-16
    '#117A65',  # ViT-32
    '#1F618D',  # Swin V2
    '#6C3483',  # CrossViT
    '#B03A2E',  # MaxViT
    '#D35400',  # DINOv2
]

palette = dict(
    zip(models, colors)
)


# ─────────────────────────────────────────
# PLOT
# ─────────────────────────────────────────

plt.rcParams[
    "font.family"
] = "Times New Roman"

plt.rcParams.update({
    'font.size': 20
})

plt.figure(
    figsize=(16, 8)
)

sns.barplot(
    data=df_idx,
    x="Dataset",
    y="ResourceIndex",
    hue="Model",
    palette=palette,
    edgecolor='white',
    linewidth=0.4
)

plt.title(
    "Combined Resource Index Across Datasets",
    fontsize=25
)

plt.ylabel(
    "Resource Index (0–1)",
    fontsize=25
)

plt.xlabel(
    "Dataset",
    fontsize=25
)

legend = plt.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, 0.99),
    ncol=6,
    frameon=True,
    fontsize=16
)

legend.get_frame().set_edgecolor('black')

legend.get_frame().set_linewidth(0.5)

plt.ylim(0,1.2)

plt.tight_layout()

plt.savefig(
    "resource_index_updated.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()


# ─────────────────────────────────────────
# PRINT RESOURCE INDEX VALUES
# ─────────────────────────────────────────

print(
    df_idx[
        [
            "Dataset",
            "Model",
            "ResourceIndex"
        ]
    ].to_string(index=False)
)


# ─────────────────────────────────────────
# LOWEST RESOURCE INDEX PER DATASET
# ─────────────────────────────────────────

print(
    "\nLowest Combined Resource Index:"
)

for d in datasets:

    subset = df_idx[
        df_idx["Dataset"] == d
    ]

    best = subset.loc[
        subset["ResourceIndex"].idxmin()
    ]

    print(
        f"{d:18s} -> "
        f"{best['Model']:15s} "
        f"({best['ResourceIndex']:.3f})"
    )
```

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
from PIL import Image

# Load the training histories extracted directly from the 11 notebook outputs.
with open("/mnt/data/training_histories_extracted.json", "r", encoding="utf-8") as f:
    histories = json.load(f)

datasets = ["WildDeepfake", "Celeb-DF (V2)", "DFC", "FF++"]

pretrained_models = [
    "XceptionNet",
    "ResNet-50",
    "Inception-v3",
    "EfficientNetV2-B0",
    "ViT-16",
    "ViT-32",
    "Swin V2",
    "CrossViT",
    "MaxViT",
    "DINOv2",
]

# A Times-compatible serif font that is available in this runtime.
plt.rcParams["font.family"] = "Liberation Serif"

out_dir = Path("/mnt/data/reviewer_training_figures")
out_dir.mkdir(exist_ok=True)

markers = ["o", "s", "^", "D", "v", "P", "X", "<", ">", "*"]


def pretrained_panel(dataset, metric, ylabel, panel_letter, ylim, path):
    fig, ax = plt.subplots(figsize=(6.4, 4.6))

    for model, marker in zip(pretrained_models, markers):
        rec = histories[model][dataset]
        epochs = [r["epoch"] for r in rec]
        values = [r[metric] for r in rec]
        ax.plot(
            epochs,
            values,
            marker=marker,
            markersize=3.5,
            linewidth=1.4,
            label=model,
        )

    ax.set_title(f"({panel_letter}) {dataset}", fontsize=15)
    ax.set_xlabel("Epoch", fontsize=13)
    ax.set_ylabel(ylabel, fontsize=13)
    ax.set_xticks(range(1, 11))
    ax.set_xlim(1, 10)
    ax.set_ylim(*ylim)
    ax.tick_params(axis="both", labelsize=10)
    ax.grid(axis="both", linestyle="--", alpha=0.25)

    fig.tight_layout()
    fig.savefig(path, dpi=220, bbox_inches="tight")
    plt.close(fig)


def mesonet_panel(dataset, train_key, val_key, ylabel, panel_letter, ylim, path):
    fig, ax = plt.subplots(figsize=(5.4, 4.0))

    rec = histories["MesoNet"][dataset]
    epochs = [r["epoch"] for r in rec]
    train_values = [r[train_key] for r in rec]
    val_values = [r[val_key] for r in rec]

    ax.plot(epochs, train_values, linewidth=1.4, label="Training")
    ax.plot(epochs, val_values, linewidth=1.4, label="Validation")

    ax.set_title(f"({panel_letter}) {dataset}", fontsize=14)
    ax.set_xlabel("Epoch", fontsize=12)
    ax.set_ylabel(ylabel, fontsize=12)
    ax.set_xlim(1, 100)
    ax.set_xticks([1, 20, 40, 60, 80, 100])
    ax.set_ylim(*ylim)
    ax.tick_params(axis="both", labelsize=9)
    ax.grid(axis="both", linestyle="--", alpha=0.25)
    ax.legend(loc="best", fontsize=9, frameon=True)

    fig.tight_layout()
    fig.savefig(path, dpi=220, bbox_inches="tight")
    plt.close(fig)


def model_legend(path):
    fig = plt.figure(figsize=(12.5, 1.0))
    handles = [
        Line2D([0], [0], marker=mk, linewidth=1.4, markersize=4, label=model)
        for model, mk in zip(pretrained_models, markers)
    ]
    fig.legend(
        handles=handles,
        labels=pretrained_models,
        loc="center",
        ncol=5,
        fontsize=10,
        frameon=True,
    )
    plt.axis("off")
    fig.savefig(path, dpi=220, bbox_inches="tight", pad_inches=0.05)
    plt.close(fig)


def combine_2x2(panel_paths, legend_path, output_path):
    panels = [Image.open(p).convert("RGB") for p in panel_paths]
    legend = Image.open(legend_path).convert("RGB")

    pw = max(p.width for p in panels)
    ph = max(p.height for p in panels)

    canvas = Image.new("RGB", (2 * pw, legend.height + 2 * ph), "white")
    canvas.paste(legend, ((canvas.width - legend.width) // 2, 0))

    spots = [
        (0, legend.height),
        (pw, legend.height),
        (0, legend.height + ph),
        (pw, legend.height + ph),
    ]
    for img, (x0, y0) in zip(panels, spots):
        canvas.paste(
            img,
            (x0 + (pw - img.width) // 2, y0 + (ph - img.height) // 2),
        )

    canvas.save(output_path, dpi=(300, 300))


def combine_2x4(top_paths, bottom_paths, output_path):
    top = [Image.open(p).convert("RGB") for p in top_paths]
    bottom = [Image.open(p).convert("RGB") for p in bottom_paths]

    all_panels = top + bottom
    pw = max(p.width for p in all_panels)
    ph = max(p.height for p in all_panels)

    canvas = Image.new("RGB", (4 * pw, 2 * ph), "white")

    for col, img in enumerate(top):
        canvas.paste(
            img,
            (col * pw + (pw - img.width) // 2, (ph - img.height) // 2),
        )

    for col, img in enumerate(bottom):
        canvas.paste(
            img,
            (
                col * pw + (pw - img.width) // 2,
                ph + (ph - img.height) // 2,
            ),
        )

    canvas.save(output_path, dpi=(300, 300))


# ------------------------------------------------------------
# FIGURE 1: VALIDATION ACCURACY
# ------------------------------------------------------------
letters = ["a", "b", "c", "d"]
fig1_panels = []

for dataset, letter in zip(datasets, letters):
    p = out_dir / f"fig1_{letter}.png"
    pretrained_panel(
        dataset,
        "val_accuracy",
        "Validation Accuracy",
        letter,
        (0.50, 1.02),
        p,
    )
    fig1_panels.append(p)

legend = out_dir / "pretrained_legend.png"
model_legend(legend)

fig1 = out_dir / "Figure_1_Validation_Accuracy.png"
combine_2x2(fig1_panels, legend, fig1)


# ------------------------------------------------------------
# FIGURE 2: VALIDATION LOSS
# ------------------------------------------------------------
fig2_panels = []

for dataset, letter in zip(datasets, letters):
    p = out_dir / f"fig2_{letter}.png"
    pretrained_panel(
        dataset,
        "val_loss",
        "Validation Loss",
        letter,
        (0.0, 1.05),
        p,
    )
    fig2_panels.append(p)

fig2 = out_dir / "Figure_2_Validation_Loss.png"
combine_2x2(fig2_panels, legend, fig2)


# ------------------------------------------------------------
# FIGURE 3: MESONET TRAINING + VALIDATION
# Top row: accuracy
# Bottom row: loss
# ------------------------------------------------------------
fig3_top = []
fig3_bottom = []

for dataset, letter in zip(datasets, letters):
    acc_path = out_dir / f"fig3_{letter}_accuracy.png"
    loss_path = out_dir / f"fig3_{letter}_loss.png"

    mesonet_panel(
        dataset,
        "train_accuracy",
        "val_accuracy",
        "Accuracy",
        letter,
        (0.45, 1.02),
        acc_path,
    )
    mesonet_panel(
        dataset,
        "train_loss",
        "val_loss",
        "Loss",
        letter,
        (0.0, 1.80),
        loss_path,
    )

    fig3_top.append(acc_path)
    fig3_bottom.append(loss_path)

fig3 = out_dir / "Figure_3_MesoNet_Training_Validation.png"
combine_2x4(fig3_top, fig3_bottom, fig3)

print("Done:")
print(fig1)
print(fig2)
print(fig3)
```
